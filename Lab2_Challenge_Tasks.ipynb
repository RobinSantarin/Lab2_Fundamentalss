# Exersice 1 - Sensor Monitoring

```python
import random

# ==============================
# ACTIVITY 1: SENSOR MONITORING
# ==============================

SURNAME = "Santarin"
SEED_NUM = 12345

# Use surname and SEED_NUM to create a unique seed
seed_value = sum(ord(char) for char in SURNAME) + SEED_NUM
random.seed(seed_value)

# Generate a fixed number of sensor readings
NUMBER_OF_READINGS = 10

sensor_data = []

for _ in range(NUMBER_OF_READINGS):
    # Mostly valid readings, with some possible invalid values
    reading = random.choice([
        round(random.uniform(0, 100), 2),
        round(random.uniform(0, 100), 2),
        -10,
        110,
        "INVALID"
    ])
    sensor_data.append(reading)

# Display generated data
print("========================================")
print("       SENSOR MONITORING SYSTEM")
print("========================================")
print(f"Surname: {SURNAME}")
print(f"SEED_NUM: {SEED_NUM}")
print(f"Number of Readings: {NUMBER_OF_READINGS}")

print("\nGenerated Sensor Data:")
for i, reading in enumerate(sensor_data, start=1):
    print(f"Reading {i}: {reading}")

# ==================================
# VALIDATION
# ==================================

valid_readings = []
invalid_readings = []

for reading in sensor_data:

    try:
        value = float(reading)

        # Valid sensor range: 0 to 100
        if 0 <= value <= 100:
            valid_readings.append(value)
        else:
            invalid_readings.append(reading)

    except (ValueError, TypeError):
        invalid_readings.append(reading)

# ==================================
# CLASSIFICATION
# ==================================

classification_results = []

for value in valid_readings:

    # Operating ranges
    if 0 <= value <= 30:
        classification = "LOW"

    elif 31 <= value <= 70:
        classification = "NORMAL"

    elif 71 <= value <= 100:
        classification = "HIGH"

    classification_results.append((value, classification))

# ==================================
# DISPLAY VALID/INVALID RESULTS
# ==================================

print("\n========================================")
print("       VALID / INVALID RESULTS")
print("========================================")

print("\nValid Readings:")
if valid_readings:
    for value in valid_readings:
        print(f"{value:.2f}")
else:
    print("None")

print("\nInvalid Readings:")
if invalid_readings:
    for value in invalid_readings:
        print(value)
else:
    print("None")

# ==================================
# DISPLAY CLASSIFICATION
# ==================================

print("\n========================================")
print("       CLASSIFICATION RESULTS")
print("========================================")

if classification_results:
    for value, classification in classification_results:
        print(f"{value:.2f} -> {classification}")
else:
    print("No valid readings to classify.")

# ==================================
# SUMMARY
# ==================================

low_count = 0
normal_count = 0
high_count = 0

for value, classification in classification_results:

    if classification == "LOW":
        low_count += 1

    elif classification == "NORMAL":
        normal_count += 1

    elif classification == "HIGH":
        high_count += 1

print("\n========================================")
print("              SUMMARY")
print("========================================")

print(f"Total Readings: {len(sensor_data)}")
print(f"Valid Readings: {len(valid_readings)}")
print(f"Invalid Readings: {len(invalid_readings)}")
print(f"LOW: {low_count}")
print(f"NORMAL: {normal_count}")
print(f"HIGH: {high_count}")

print("\nExecution Log:")
print("1. Sensor data generated.")
print("2. Readings validated.")
print("3. Valid readings classified.")
print("4. Summary generated.")
print("5. Activity 1 completed.")
```

# Exersice 2 - Signal Character Analysis

```python
# ==================================
# ACTIVITY 2: SIGNAL ANALYSIS
# ==================================

SURNAME = "Santarin"
SEED_NUM = 12345

# Generate a unique signal using surname and SEED_NUM
generated_signal = f"{SURNAME}-{SEED_NUM} ECE!@#"

print("========================================")
print("        SIGNAL ANALYSIS SYSTEM")
print("========================================")

print(f"Surname: {SURNAME}")
print(f"SEED_NUM: {SEED_NUM}")

# ==================================
# GENERATED SIGNAL
# ==================================

print("\nGenerated Signal:")
print(generated_signal)

# ==================================
# PROCESS AND NORMALIZE SIGNAL
# ==================================

processed_signal = generated_signal.strip().lower()

print("\nProcessed Signal:")
print(processed_signal)

# ==================================
# CHARACTER ANALYSIS
# ==================================

letters = 0
digits = 0
spaces = 0
special_characters = 0

print("\n========================================")
print("         CHARACTER ANALYSIS")
print("========================================")

for character in processed_signal:

    if character.isalpha():
        category = "Letter"
        letters += 1

    elif character.isdigit():
        category = "Digit"
        digits += 1

    elif character.isspace():
        category = "Space"
        spaces += 1

    else:
        category = "Special Character"
        special_characters += 1

    print(f"{repr(character)} -> {category}")

# ==================================
# FINAL CLASSIFICATION
# ==================================

total_characters = len(processed_signal)

print("\n========================================")
print("        SIGNAL CLASSIFICATION")
print("========================================")

print(f"Total Characters: {total_characters}")
print(f"Letters: {letters}")
print(f"Digits: {digits}")
print(f"Spaces: {spaces}")
print(f"Special Characters: {special_characters}")

# ==================================
# DIAGNOSTIC REPORT
# ==================================

print("\n========================================")
print("          DIAGNOSTIC REPORT")
print("========================================")

if letters > 0 and digits > 0:
    signal_classification = "ALPHANUMERIC SIGNAL"
elif letters > 0:
    signal_classification = "TEXT SIGNAL"
elif digits > 0:
    signal_classification = "NUMERIC SIGNAL"
else:
    signal_classification = "SPECIAL CHARACTER SIGNAL"

print(f"Signal Classification: {signal_classification}")

# ==================================
# EXECUTION LOG
# ==================================

print("\nExecution Log:")
print("1. Signal generated using surname and SEED_NUM.")
print("2. Signal processed and normalized.")
print("3. Signal analyzed character by character.")
print("4. Diagnostic report generated.")
print("5. Activity 2 completed.")
```

# Activity 3 - Password Authentication

```python
# ==================================
# ACTIVITY 3: PASSWORD AUTHENTICATION
# ==================================

SURNAME = "Santarin"
SEED_NUM = 12345

# Generate unique password using surname and SEED_NUM
generated_password = f"{SURNAME}{SEED_NUM}"

# Attempt limit
attempt_limit = 3

attempts_made = 0
access_granted = False

print("========================================")
print("      ELECTRONIC ACCESS SYSTEM")
print("========================================")

print(f"Surname: {SURNAME}")
print(f"SEED_NUM: {SEED_NUM}")
print(f"Attempt Limit: {attempt_limit}")

# ==================================
# AUTHENTICATION PROCESS
# ==================================

print("\nAuthentication started.")

while attempts_made < attempt_limit:

    entered_password = input("Enter password: ")

    attempts_made += 1

    if entered_password == generated_password:

        access_granted = True

        print("\nAccess Granted!")
        print("Authentication successful.")

        break

    else:

        remaining_attempts = attempt_limit - attempts_made

        print("\nIncorrect Password.")

        if remaining_attempts > 0:
            print(f"Attempts remaining: {remaining_attempts}")
        else:
            print("Maximum number of attempts reached.")
            print("Authentication terminated.")

# ==================================
# FINAL SYSTEM STATE
# ==================================

if access_granted:

    access_result = "GRANTED"
    final_system_state = "SYSTEM UNLOCKED"

else:

    access_result = "DENIED"
    final_system_state = "SYSTEM LOCKED"

# ==================================
# ASSESSMENT DATA
# ==================================

print("\n========================================")
print("          ASSESSMENT DATA")
print("========================================")

print(f"Generated Password: {generated_password}")
print(f"Attempt Limit: {attempt_limit}")
print(f"Attempts Made: {attempts_made}")
print(f"Access Result: {access_result}")
print(f"Final System State: {final_system_state}")

# ==================================
# EXECUTION LOG
# ==================================

print("\n========================================")
print("           EXECUTION LOG")
print("========================================")

if access_granted:

    print("1. Password generated.")
    print("2. Authentication attempt received.")
    print("3. Correct password entered.")
    print("4. Access granted.")
    print("5. System unlocked.")

else:

    print("1. Password generated.")
    print("2. Authentication attempts received.")
    print("3. Maximum attempt limit reached.")
    print("4. Access denied.")
    print("5. System locked.")

print("\nActivity 3 completed.")
```
